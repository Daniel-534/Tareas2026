## 1) Función suma de divisores $\sigma(n)$

```python
def sigma(n: int) -> int:
    """Retorna la suma de todos los divisores positivos de n."""
    if n <= 0:
        raise ValueError("n debe ser un entero positivo")

    suma_divisores = 0
    for divisor in range(1, n + 1):
        if n % divisor == 0:
            suma_divisores += divisor
    return suma_divisores

# Ejemplo rápido
print("sigma(6) =", sigma(6))
```

## 2) Funciones para clasificar números

```python
def es_deficiente(n: int) -> bool:
    """Retorna True si n es deficiente: sigma(n) < 2n."""
    return sigma(n) < 2 * n


def es_perfecto(n: int) -> bool:
    """Retorna True si n es perfecto: sigma(n) = 2n."""
    return sigma(n) == 2 * n


def es_abundante(n: int) -> bool:
    """Retorna True si n es abundante: sigma(n) > 2n."""
    return sigma(n) > 2 * n


for numero in [8, 28, 12]:
    print(
        numero,
        "deficiente=", es_deficiente(numero),
        "perfecto=", es_perfecto(numero),
        "abundante=", es_abundante(numero),
    )
```

## 3) ¿Por qué reciben esos nombres?

La clasificación compara la **suma de divisores** con el valor de referencia $2n$:

- **Deficiente**: la suma queda por debajo de $2n$ ("falta" cantidad).
- **Perfecto**: la suma coincide exactamente con $2n$ (equilibrio exacto).
- **Abundante**: la suma supera $2n$ ("sobra" cantidad).

## 4) Scatter plot de $(n, \sigma(n))$

Se colorea cada punto según su tipo:
- Rojo: deficiente
- Verde: perfecto
- Azul: abundante

```python
import matplotlib.pyplot as plt

n_min, n_max = 1, 120
valores_n = list(range(n_min, n_max + 1))
sigma_por_n = {n: sigma(n) for n in valores_n}

n_deficientes = [n for n in valores_n if sigma_por_n[n] < 2 * n]
n_perfectos = [n for n in valores_n if sigma_por_n[n] == 2 * n]
n_abundantes = [n for n in valores_n if sigma_por_n[n] > 2 * n]

sigma_deficientes = [sigma_por_n[n] for n in n_deficientes]
sigma_perfectos = [sigma_por_n[n] for n in n_perfectos]
sigma_abundantes = [sigma_por_n[n] for n in n_abundantes]

plt.figure(figsize=(10, 6))
plt.scatter(n_deficientes, sigma_deficientes, color="red", s=25, label="Deficientes")
plt.scatter(n_perfectos, sigma_perfectos, color="green", s=50, label="Perfectos")
plt.scatter(n_abundantes, sigma_abundantes, color="blue", s=25, label="Abundantes")

# Referencia sigma(n) = 2n
plt.plot(valores_n, [2 * n for n in valores_n], color="black", linestyle="--", label="sigma(n) = 2n")

plt.title("Clasificación de números por sigma(n)")
plt.xlabel("n")
plt.ylabel("sigma(n)")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

## 5) Reflexión después del gráfico

El gráfico confirma visualmente la regla de comparación con la recta $\sigma(n)=2n$:

- Los **deficientes** quedan debajo de la recta.
- Los **perfectos** caen exactamente sobre la recta.
- Los **abundantes** quedan por encima.

Comparado con la respuesta del punto 3, la idea es la misma, pero ahora con evidencia visual: los nombres describen la posición relativa de $\sigma(n)$ frente a $2n$.

## 6) Cinco preguntas de estadística sobre el análisis

Se proponen 2 preguntas fáciles, 2 medias y 1 difícil, cada una con cálculo y respuesta.

```python
from statistics import mean, median

n_min, n_max = 1, 200
universo = list(range(n_min, n_max + 1))
sigmas = {n: sigma(n) for n in universo}

def clasificacion(n: int) -> str:
    if es_deficiente(n):
        return "deficiente"
    if es_perfecto(n):
        return "perfecto"
    return "abundante"

clases = {n: clasificacion(n) for n in universo}
deficientes = [n for n in universo if clases[n] == "deficiente"]
perfectos = [n for n in universo if clases[n] == "perfecto"]
abundantes = [n for n in universo if clases[n] == "abundante"]

# -------------------------
# FÁCIL 1
# -------------------------
p_deficientes = len(deficientes) / len(universo)
print("Fácil 1) Proporción de deficientes en [1,200]:", round(p_deficientes, 4))

# -------------------------
# FÁCIL 2
# -------------------------
media_sigma = mean(sigmas.values())
mediana_sigma = median(sigmas.values())
print("Fácil 2) Media de sigma(n):", round(media_sigma, 2))
print("          Mediana de sigma(n):", mediana_sigma)

# -------------------------
# MEDIO 1
# -------------------------
media_sigma_def = mean(sigmas[n] for n in deficientes)
media_sigma_abu = mean(sigmas[n] for n in abundantes)
print("Medio 1) Media sigma(n) en deficientes:", round(media_sigma_def, 2))
print("         Media sigma(n) en abundantes:", round(media_sigma_abu, 2))

# -------------------------
# MEDIO 2
# -------------------------
pares = [n for n in universo if n % 2 == 0]
impares = [n for n in universo if n % 2 == 1]
p_abu_par = len([n for n in pares if clases[n] == "abundante"]) / len(pares)
p_abu_impar = len([n for n in impares if clases[n] == "abundante"]) / len(impares)
print("Medio 2) P(abundante | par):", round(p_abu_par, 4))
print("         P(abundante | impar):", round(p_abu_impar, 4))

# -------------------------
# DIFÍCIL 1
# -------------------------
# Correlación de Pearson entre n y sigma(n)
x = universo
y = [sigmas[n] for n in universo]
mx, my = mean(x), mean(y)
cov_xy = sum((xi - mx) * (yi - my) for xi, yi in zip(x, y)) / len(x)
var_x = sum((xi - mx) ** 2 for xi in x) / len(x)
var_y = sum((yi - my) ** 2 for yi in y) / len(y)
corr = cov_xy / ((var_x ** 0.5) * (var_y ** 0.5))
print("Difícil 1) Correlación Pearson entre n y sigma(n):", round(corr, 4))
```

### Interpretación breve de las 5 preguntas

1. **Fácil 1:** cuantifica qué tan frecuentes son los deficientes en el rango.
2. **Fácil 2:** compara tendencia central (media vs mediana) de $\sigma(n)$.
3. **Medio 1:** contrasta el comportamiento promedio de $\sigma(n)$ entre deficientes y abundantes.
4. **Medio 2:** evalúa si la paridad influye en la abundancia.
5. **Difícil 1:** mide relación lineal global entre $n$ y $\sigma(n)$ mediante correlación de Pearson.
